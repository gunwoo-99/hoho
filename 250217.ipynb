```python
from sklearn.cluster import KMeans

# Elbow Point를 찾기 위한 for문
x = [] # k가 몇개인지
y = [] # 응집도가 몇인지

for k in range(1,5) :
    kmeans = KMeans(n_clusters = k, n_init="auto")
    kmeans.fit(df1_d)
    
    x.append(k)
    y.append(kmeans.inertia_)
    
# 그래프 그리기
plt.plot(x, y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Elbow Method를 위한 변수
x = []  # k 값 (클러스터 개수)
y = []  # 음의 로그 가능도 (Negative Log-Likelihood)

# K값을 1~5까지 변화시키며 모델 학습
for k in range(1, 6):  # 기존 range(1,5)에서 range(1,6)으로 확장
    gm = GaussianMixture(n_components=k, random_state=0)
    gm.fit(df1_d)

    # GaussianMixture에서는 inertia_ 대신 log-likelihood 사용
    log_likelihood = gm.score(df1_d) * len(df1_d)  # 전체 로그 가능도 계산
    nll = -log_likelihood  # Elbow Method를 위해 음수 변환

    x.append(k)
    y.append(nll)

# 그래프 그리기
plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.xlabel("Number of Components (K)")
plt.ylabel("Negative Log-Likelihood")
plt.title("Elbow Method for Optimal K (Gaussian Mixture)")
plt.grid()
plt.show()
```

```python
from sklearn.mixture import GaussianMixture

# 클러스터 개수 설정
k = 5

# Gaussian Mixture 모델 생성 및 학습
gmm = GaussianMixture(n_components=k, random_state=42)
gmm.fit(df1_d)

# 클러스터 할당 (각 샘플이 속하는 클러스터 라벨)
df1_d['cluster'] = gmm.predict(df1_d)

# 클러스터 할당 확률 확인 (선택적)
df1_d['cluster_prob'] = gmm.predict_proba(df1_d).max(axis=1)

print(df1_d.head())
```
